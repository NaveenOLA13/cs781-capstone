```python
import os, subprocess, sys, glob
from kaggle_secrets import UserSecretsClient

GH = UserSecretsClient().get_secret("GH_TOKEN")
REPO = "/kaggle/working/cs781-capstone"
if not os.path.isdir(REPO):
    assert (
        subprocess.run(
            ["git", "clone", "-q", f"https://{GH}@github.com/NaveenOLA13/cs781-capstone.git", REPO]
        ).returncode
        == 0
    )
sys.path.insert(0, f"{REPO}/phase5")
bases = sorted(set(glob.glob("/kaggle/input/**/lsi_runs", recursive=True)))
print("result folders found:", *bases, sep="\n  ")
assert bases, "attach the four notebooks' outputs (or the results zips) first"
```

```python
from lsi_grid import ensemble, report

final = ensemble(bases, out_dir="/kaggle/working/lsi_final", max_members=8)
report(bases + ["/kaggle/working/lsi_final"], out_path="/kaggle/working/all_results.txt")
```

```python
import zipfile
from IPython.display import FileLink, display

print("members:", final["members"])
display(FileLink("lsi_final/submission/submission.zip"))
with zipfile.ZipFile("/kaggle/working/lsi_final_results.zip", "w", zipfile.ZIP_DEFLATED) as z:
    for root, _, files in os.walk("/kaggle/working/lsi_final"):
        for f in files:
            z.write(os.path.join(root, f), os.path.relpath(os.path.join(root, f), "/kaggle/working"))
    z.write("/kaggle/working/all_results.txt", "all_results.txt")
display(FileLink("lsi_final_results.zip"))
```
